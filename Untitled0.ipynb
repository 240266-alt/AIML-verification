```python
class SimpleANN(nn.Module):
    def __init__(self, input_dim, hidden_dim):
        super(SimpleANN, self).__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.fc1(x)
        out = self.relu(out)
        out = self.fc2(out)
        out = self.sigmoid(out)
        return out

model = SimpleANN(input_dim=30, hidden_dim=32)
print(model)
```

```python
class NumPy2LayerNet:
    def __init__(self, input_dim, hidden_dim, output_dim=1, seed=42):
        np.random.seed(seed)

        self.W1 = np.random.randn(input_dim, hidden_dim) * np.sqrt(2.0 / input_dim)
        self.b1 = np.zeros((1, hidden_dim))

        self.W2 = np.random.randn(hidden_dim, output_dim) * np.sqrt(1.0 / hidden_dim)
        self.b2 = np.zeros((1, output_dim))

    def sigmoid(self, z):
        return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))

    def relu(self, z):
        return np.maximum(0, z)

    def forward(self, X):
        self.X = X
        self.Z1 = np.dot(X, self.W1) + self.b1
        self.A1 = self.relu(self.Z1)
        self.Z2 = np.dot(self.A1, self.W2) + self.b2
        self.A2 = self.sigmoid(self.Z2)
        return self.A2

    def backward(self, Y):
        m = self.X.shape[0]


        dZ2 = self.A2 - Y
        dW2 = np.dot(self.A1.T, dZ2) / m
        db2 = np.sum(dZ2, axis=0, keepdims=True) / m


        dA1 = np.dot(dZ2, self.W2.T)
        dZ1 = dA1 * (self.Z1 > 0)
        dW1 = np.dot(self.X.T, dZ1) / m
        db1 = np.sum(dZ1, axis=0, keepdims=True) / m

        return dW1, db1, dW2, db2

    def get_loss(self, A2, Y):
        m = Y.shape[0]
        epsilon = 1e-15
        A2 = np.clip(A2, epsilon, 1 - epsilon)
        return -np.sum(Y * np.log(A2) + (1 - Y) * np.log(1 - A2)) / m
```

### Module 9: Numerical Gradient Checking

We verify our analytically computed gradients against a numerical finite-difference approximation to ensure mathematical correctness.

```python
def numerical_gradient_check():

    X_check, y_check = X_train[:5], y_train[:5].reshape(-1, 1)
    net = NumPy2LayerNet(input_dim=30, hidden_dim=4, seed=10)


    net.forward(X_check)
    dW1, db1, dW2, db2 = net.backward(y_check)


    epsilon = 1e-5
    for param_name, param, grad in [('W2', net.W2, dW2), ('W1', net.W1, dW1)]:
        num_grad = np.zeros_like(param)
        it = np.nditer(param, flags=['multi_index'], op_flags=['readwrite'])
        while not it.finished:
            idx = it.multi_index
            old_val = param[idx]

            param[idx] = old_val + epsilon
            loss_plus = net.get_loss(net.forward(X_check), y_check)

            param[idx] = old_val - epsilon
            loss_minus = net.get_loss(net.forward(X_check), y_check)

            num_grad[idx] = (loss_plus - loss_minus) / (2 * epsilon)
            param[idx] = old_val
            it.iternext()


        rel_error = np.linalg.norm(grad - num_grad) / (np.linalg.norm(grad) + np.linalg.norm(num_grad) + 1e-15)
        print(f"Relative Error for parameter {param_name}: {rel_error:.2e}")

numerical_gradient_check()
```

```python
architectures = [16, 32, 64]
learning_rates = [0.01, 0.05, 0.1]
epochs = 300

plt.figure(figsize=(12, 8))
results = {}


y_train_np = y_train.reshape(-1, 1)
y_test_np = y_test.reshape(-1, 1)

for h_dim in architectures:
    for lr in learning_rates:
        net = NumPy2LayerNet(input_dim=30, hidden_dim=h_dim)
        train_losses = []
        val_losses = []

        for epoch in range(epochs):

            train_preds = net.forward(X_train)
            loss = net.get_loss(train_preds, y_train_np)
            train_losses.append(loss)


            val_preds = net.forward(X_test)
            val_loss = net.get_loss(val_preds, y_test_np)
            val_losses.append(val_loss)


            net.forward(X_train)
            dW1, db1, dW2, db2 = net.backward(y_train_np)

            net.W1 -= lr * dW1
            net.b1 -= lr * db1
            net.W2 -= lr * dW2
            net.b2 -= lr * db2

        label = f'Arch: {h_dim} nodes, LR: {lr}'
        results[label] = {
            'net': net,
            'train_losses': train_losses,
            'val_losses': val_losses,
            'final_val_loss': val_losses[-1]
        }
        plt.plot(train_losses, label=label)

plt.title('NumPy Scratch ANN: Comparative Learning Curves')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.yscale('log')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, which='both', linestyle='--')
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc, precision_recall_curve, average_precision_score


best_label = min(results, key=lambda k: results[k]['final_val_loss'])
print(f"Best configuration: {best_label}")
best_net = results[best_label]['net']


test_probabilities = best_net.forward(X_test)

fpr, tpr, thresholds_roc = roc_curve(y_test_np, test_probabilities)
roc_auc = auc(fpr, tpr)

precisions, recalls, thresholds_pr = precision_recall_curve(y_test_np, test_probabilities)
ap_score = average_precision_score(y_test_np, test_probabilities)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))


axes[0].plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.4f})')
axes[0].plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
axes[0].set_xlabel('False Positive Rate')
axes[0].set_ylabel('True Positive Rate')
axes[0].set_title('Receiver Operating Characteristic')
axes[0].legend(loc="lower right")
axes[0].grid(True)


axes[1].plot(recalls, precisions, color='green', lw=2, label=f'PR Curve (AP = {ap_score:.4f})')
axes[1].set_xlabel('Recall')
axes[1].set_ylabel('Precision')
axes[1].set_title('Precision-Recall Curve')
axes[1].legend(loc="lower left")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt


default_preds = (test_probabilities >= 0.5).astype(int)
cm_before = confusion_matrix(y_test_np, default_preds)


max_recall = np.max(recalls)
high_recall_idx = np.where(recalls == max_recall)[0]


valid_indices = high_recall_idx[high_recall_idx < len(thresholds_pr)]
best_idx = valid_indices[-1]
optimized_threshold = thresholds_pr[best_idx]
print(f"Optimized Decision Threshold (prioritizing Recall): {optimized_threshold:.6f}")

custom_preds = (test_probabilities >= optimized_threshold).astype(int)
cm_after = confusion_matrix(y_test_np, custom_preds)


fig, axes = plt.subplots(1, 2, figsize=(12, 5))
disp1 = ConfusionMatrixDisplay(confusion_matrix=cm_before, display_labels=['Benign', 'Malignant'])
disp1.plot(ax=axes[0], cmap=plt.cm.Oranges, values_format='d')
axes[0].set_title("Before Adjustment (Threshold = 0.5)")

disp2 = ConfusionMatrixDisplay(confusion_matrix=cm_after, display_labels=['Benign', 'Malignant'])
disp2.plot(ax=axes[1], cmap=plt.cm.Greens, values_format='d')
axes[1].set_title(f"After Adjustment (Threshold = {optimized_threshold:.6f})")

plt.tight_layout()
plt.show()
```
