```python
import pandas as pd
from collections import defaultdict, deque
import time
import os

# ============================================================
# 1. LOAD DATASET
# ============================================================

import kagglehub

# Download latest version
path = kagglehub.dataset_download("bhavikjikadara/us-airline-flight-routes-and-fares-1993-2024")

print("Path to dataset files:", path)

# Construct the full file path to the CSV file
file_name = "US Airline Flight Routes and Fares 1993-2024.csv"
full_file_path = os.path.join(path, file_name)
df = pd.read_csv(full_file_path, low_memory=False)

print("=" * 60)
print("US AIRLINE FLIGHT ROUTES AND FARES DATASET")
print("=" * 60)

print("\nDataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 2. CREATE GRAPH
# ============================================================

# Remove rows with missing values
routes = df.dropna(
    subset=["city1", "city2", "nsmiles"]
).copy()

# Keep only valid distances
routes = routes[routes["nsmiles"] > 0]

# Graph:
# city -> [(neighbor, distance), ...]
graph = defaultdict(list)

for _, row in routes.iterrows():

    city1 = row["city1"]
    city2 = row["city2"]
    distance = float(row["nsmiles"])

    # Add both directions
    graph[city1].append((city2, distance))
    graph[city2].append((city1, distance))


print("\nNumber of cities:", len(graph))
print("Number of routes:", len(routes))


# ============================================================
# 3. DISPLAY SOME AVAILABLE CITIES
# ============================================================

print("\nSample cities available in dataset:")

cities = sorted(graph.keys())

for i, city in enumerate(cities[:30], start=1):
    print(i, ".", city)


# ============================================================
# 4. SELECT SOURCE AND DESTINATION
# ============================================================

# Change these according to the cities in your dataset

start = "New York, NY"
goal = "Los Angeles, CA"

# Check whether cities exist
if start not in graph:
    print("\nERROR:", start, "is not available in the dataset.")
    print("Choose a city from the list above.")
    exit()

if goal not in graph:
    print("\nERROR:", goal, "is not available in the dataset.")
    print("Choose a city from the list above.")
    exit()


# ============================================================
# 5. GREEDY SEARCH
# ============================================================

def greedy_search(graph, start, goal):

    current = start

    path = [current]

    visited = {current}

    total_distance = 0

    nodes_explored = 0

    while current != goal:

        nodes_explored += 1

        # Get unvisited neighbors
        neighbors = [
            (city, distance)
            for city, distance in graph[current]
            if city not in visited
        ]

        # No possible route
        if not neighbors:
            return None, None, nodes_explored

        # Greedy choice:
        # Choose the nearest city
        next_city, distance = min(
            neighbors,
            key=lambda x: x[1]
        )

        path.append(next_city)

        total_distance += distance

        visited.add(next_city)

        current = next_city

    nodes_explored += 1

    return path, total_distance, nodes_explored


# ============================================================
# 6. BFS
# ============================================================

def bfs(graph, start, goal):

    queue = deque()

    queue.append([start])

    visited = {start}

    nodes_explored = 0

    while queue:

        path = queue.popleft()

        current = path[-1]

        nodes_explored += 1

        # Goal found
        if current == goal:
            return path, nodes_explored

        # Explore neighbors
        for neighbor, distance in graph[current]:

            if neighbor not in visited:

                visited.add(neighbor)

                new_path = path + [neighbor]

                queue.append(new_path)

    return None, nodes_explored


# ============================================================
# 7. DFS
# ============================================================

def dfs(graph, start, goal):

    stack = []

    stack.append([start])

    visited = set()

    nodes_explored = 0

    while stack:

        path = stack.pop()

        current = path[-1]

        if current in visited:
            continue

        visited.add(current)

        nodes_explored += 1

        # Goal found
        if current == goal:
            return path, nodes_explored

        # Add neighbors to stack
        for neighbor, distance in graph[current]:

            if neighbor not in visited:

                new_path = path + [neighbor]

                stack.append(new_path)

    return None, nodes_explored


# ============================================================
# 8. RUN GREEDY SEARCH
# ============================================================

print("\n" + "=" * 60)
print("GREEDY SEARCH")
print("=" * 60)

start_time = time.perf_counter()

greedy_path, greedy_distance, greedy_nodes = greedy_search(
    graph,
    start,
    goal
)

greedy_time = time.perf_counter() - start_time

if greedy_path:

    print("\nRoute:")
    print(" -> ".join(greedy_path))

    print("\nNumber of connections:",
          len(greedy_path) - 1)

    print("Total distance:",
          round(greedy_distance, 2),
          "miles")

    print("Nodes explored:",
          greedy_nodes)

    print("Execution time:",
          round(greedy_time, 6),
          "seconds")

else:

    print("No route found.")


# ============================================================
# 9. RUN BFS
# ============================================================

print("\n" + "=" * 60)
print("BREADTH-FIRST SEARCH (BFS)")
print("=" * 60)

start_time = time.perf_counter()

bfs_path, bfs_nodes = bfs(
    graph,
    start,
    goal
)

bfs_time = time.perf_counter() - start_time

if bfs_path:

    # Calculate actual distance of BFS route
    bfs_distance = 0

    for i in range(len(bfs_path) - 1):

        current = bfs_path[i]
        next_city = bfs_path[i + 1]

        for neighbor, distance in graph[current]:

            if neighbor == next_city:
                bfs_distance += distance
                break

    print("\nRoute:")
    print(" -> ".join(bfs_path))

    print("\nNumber of connections:",
          len(bfs_path) - 1)

    print("Total distance:",
          round(bfs_distance, 2),
          "miles")

    print("Nodes explored:",
          bfs_nodes)

    print("Execution time:",
          round(bfs_time, 6),
          "seconds")

else:

    print("No route found.")


# ============================================================
# 10. RUN DFS
# ============================================================

print("\n" + "=" * 60)
print("DEPTH-FIRST SEARCH (DFS)")
print("=" * 60)

start_time = time.perf_counter()

dfs_path, dfs_nodes = dfs(
    graph,
    start,
    goal
)

dfs_time = time.perf_counter() - start_time

if dfs_path:

    # Calculate distance of DFS route
    dfs_distance = 0

    for i in range(len(dfs_path) - 1):

        current = dfs_path[i]
        next_city = dfs_path[i + 1]

        for neighbor, distance in graph[current]:

            if neighbor == next_city:
                dfs_distance += distance
                break

    print("\nRoute:")
    print(" -> ".join(dfs_path))

    print("\nNumber of connections:",
          len(dfs_path) - 1)

    print("Total distance:",
          round(dfs_distance, 2),
          "miles")

    print("Nodes explored:",
          dfs_nodes)

    print("Execution time:",
          round(dfs_time, 6),
          "seconds")

else:

    print("No route found.")


# ============================================================
# 11. COMPARISON
# ============================================================

print("\n" + "=" * 60)
print("ALGORITHM COMPARISON")
print("=" * 60)

comparison = []

if greedy_path:
    comparison.append([
        "Greedy",
        len(greedy_path) - 1,
        round(greedy_distance, 2),
        greedy_nodes,
        round(greedy_time, 6)
    ])

if bfs_path:
    comparison.append([
        "BFS",
        len(bfs_path) - 1,
        round(bfs_distance, 2),
        bfs_nodes,
        round(bfs_time, 6)
    ])

if dfs_path:
    comparison.append([
        "DFS",
        len(dfs_path) - 1,
        round(dfs_distance, 2),
        dfs_nodes,
        round(dfs_time, 6)
    ])


comparison_df = pd.DataFrame(
    comparison,
    columns=[
        "Algorithm",
        "Connections",
        "Distance (miles)",
        "Nodes Explored",
        "Time (seconds)"
    ]
)

print("\n")
print(comparison_df.to_string(index=False))


# ============================================================
# 12. FINAL CONCLUSION
# ============================================================

print("\n" + "=" * 60)
print("CONCLUSION")
print("=" * 60)

print("""
Greedy:
- Selects the locally shortest route at every step.
- Does not guarantee the globally shortest route.

BFS:
- Finds the route with the minimum number of connections.
- Does not consider distance.

DFS:
- Explores one route deeply before backtracking.
- Does not guarantee the shortest route.

Therefore:
- Use Greedy when a locally optimal decision is useful.
- Use BFS when minimum number of connections is important.
- Use DFS when complete/deep graph exploration is required.
""")
```
